## Bloque 1: Antidumping aplicado por Argentina

Argentina es el **5to aplicador mundial** de medidas antidumping (462 medidas iniciadas entre 1995 y 2023). Vamos a analizar contra que países se dirigen esas medidas.


```python
# Cargar datos WTO sobre orígenes de medidas AD argentinas
arg_ad = pd.read_csv(f'{DATOS}/wto_antidumping_argentina_origen.csv', comment='#')
print('Total medidas AD iniciadas por Argentina:', arg_ad['medidas'].sum())
print()
arg_ad
```

```python
# Visualización: barras horizontales
fig, ax = plt.subplots(figsize=(11, 5))
data = arg_ad.sort_values('medidas', ascending=True)
colores = ['#E74C3C' if c == 'China' else '#2E75B6' for c in data['origen']]
ax.barh(data['origen'], data['medidas'], color=colores, edgecolor='white')

total = data['medidas'].sum()
for i, (país, m) in enumerate(zip(data['origen'], data['medidas'])):
    pct = m / total * 100
    ax.text(m + 3, i, f'{m} ({pct:.0f}%)', va='center', fontsize=9)

ax.set_xlim(0, 270)
ax.set_xticks([])
ax.spines['left'].set_visible(False)
ax.set_title('Medidas AD iniciadas por Argentina, por país de origen (1995-2023)',
             fontsize=12, fontweight='bold', loc='left', pad=10)
ax.set_xlabel('Medidas iniciadas')
plt.tight_layout()
plt.show()
```

### Tarea: Agrupar por región

**Objetivo**: agrupar las medidas argentinas en tres regiones (Asia, Mercosur, Resto) y calcular cuánto pesa cada una.

Pueden usar el código siguiente o generarlo con IA (Gemini, ChatGPT) — lo importante es **interpretar el resultado**.


```python
# Definir regiones
asia = ['China', 'Korea, Rep. of', 'Chinese Taipei', 'India', 'Indonesia', 'Vietnam', 'Thailand']
mercosur = ['Brazil']

def clasificar(origen):
    if origen in asia:
        return 'Asia'
    elif origen in mercosur:
        return 'Mercosur'
    else:
        return 'Resto'

arg_ad_grupo = arg_ad.copy()
arg_ad_grupo['region'] = arg_ad_grupo['origen'].apply(clasificar)
por_region = arg_ad_grupo.groupby('region')['medidas'].sum().sort_values(ascending=False)

total = por_region.sum()
for region, m in por_region.items():
    pct = m / total * 100
    print(f'{region:10s}: {m:3d} medidas ({pct:.1f}%)')
```

## Bloque 2: Comercio intraindustrial Argentina-Brasil por sector

En la clase vimos que el comercio bilateral Argentina-Brasil en vehículos tiene un GL = 0,92 (muy alto). Pero el patrón varia mucho entre sectores. Vamos a calcular el GL para 10 capítulos HS distintos y ver que sectores tienen IIT alto vs bajo.

**Recuerden la fórmula del Grubel-Lloyd**:

$$GL_i = 1 - \frac{|X_i - M_i|}{X_i + M_i}$$

Donde $X_i$ son las exportaciones del país hacia el socio en el sector $i$, y $M_i$ las importaciones desde el socio en el mismo sector. GL en [0, 1]: cerca de 0 = comercio interindustrial (Ricardo/H-O), cerca de 1 = intraindustrial (Krugman).


```python
# Cargar datos bilaterales ARG-BRA por sector
iit = pd.read_csv(f'{DATOS}/comercio_arg_bra_por_sector_2023.csv', comment='#')
print('Sectores cargados:', len(iit))
iit
```

```python
# Calcular el GL para cada sector
iit['gl'] = (1 - abs(iit['exports_arg_a_bra_musd'] - iit['imports_arg_desde_bra_musd']) /
                 (iit['exports_arg_a_bra_musd'] + iit['imports_arg_desde_bra_musd'])).round(3)

def clasificar_iit(gl):
    if gl >= 0.7:
        return 'IIT alto (Krugman)'
    elif gl >= 0.3:
        return 'IIT medio'
    else:
        return 'IIT bajo (Ricardo/H-O)'

iit['tipo'] = iit['gl'].apply(clasificar_iit)
iit_sorted = iit.sort_values('gl', ascending=False)
iit_sorted[['sector', 'exports_arg_a_bra_musd', 'imports_arg_desde_bra_musd', 'gl', 'tipo']]
```
